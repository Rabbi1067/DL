```python
# import library and module
import tensorflow as tf
from tensorflow import keras
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('PotatoPrice - PotatoPrice.csv')
```

```python
X = df[['potato_kg']]  #since x have to be two dimentional or 2D array. So [[]]
Y = df['price']
```

```python
X
```

```python
# Split Dataset for Train and Test
from sklearn.model_selection import train_test_split

X_train,X_test,Y_train,Y_test = train_test_split(X,Y,test_size=0.3)
# X_train
```

```python
X_train
```

```python
# ANN Model Building
'''
model = tf.keras.Sequential([
      tf.keras.layers.Dense(1,input_shape=[1],activation='linear')
])

# model.summary()'''
model = tf.keras.Sequential([
    tf.keras.layers.Dense(10, activation='relu', input_shape=[1]),
    tf.keras.layers.Dense(1)
])
```

```python
# Compile The ANN model
'''model.compile(optimizer='sgd', #adam-weight update kore
              loss='mean_squared_error',
             metrics=['accuracy'])'''
model.compile(optimizer='adam', loss='mse', metrics=['mae'])
```

```python
#Train ANN Model
model.fit(X_train,Y_train,epochs=500)
```

```python
X_test
```

```python
#Predict
model.predict(X_test)
# X_test
```

```python
#Predict The Output Manually
model.predict(np.array([[4]]))
```
